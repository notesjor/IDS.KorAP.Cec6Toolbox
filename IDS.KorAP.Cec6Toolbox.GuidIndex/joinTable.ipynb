```python
import pandas as pd
import json

basedir = "C:/Users/Jan/Desktop/"

excel = pd.read_excel(basedir + 'excel.xlsx', sheet_name='data')
meta = pd.read_csv(basedir + 'meta.tsv', sep='\t', header=None)

meta.columns = ['GUID', 'JSON'] # rename columns
meta['GUID'] = meta['GUID'].str.replace('-', '') # remove dashes from GUIDs

meta_dict = meta.set_index('GUID').T.to_dict('list')

meta_dict = {k: json.loads(v[0]) for k, v in meta_dict.items()}

excel['Korpus'] = excel['GUID'].map(lambda x: meta_dict[x]['Korpus'])
excel['Sigle'] = excel['GUID'].map(lambda x: meta_dict[x]['Sigle'])
excel['Titel (D)'] = excel['GUID'].map(lambda x: meta_dict[x]['Titel (D)'])
print(excel)
```
